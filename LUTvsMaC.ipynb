```python
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
import pandas as pd
```

```python
original_path = '/mnt/ssd/bojing/Image-Adaptive-3DLUT/data/Batch_1/tiles/'
LUT_path = '/mnt/ssd/ferbue/Image-Adaptive-3DLUT/LUTs/unpaired/exp_20/evaluation/best_fid_png/'
macenko_path = '/mnt/ssd/bojing/Image-Adaptive-3DLUT/data/Batch_1/macenko_normalisation/tiles_macenko/'

slide = 'SCANB_MALMO_RiDn0435_slide_657348_WSI_clean' #Identified Macenko failure --> Largest euclidean distance H vs E, weird colors

# all_tiles = glob.glob(LUT_path+wsi_name+'/*.png')
all_tiles = glob.glob(original_path+slide+'/'+slide+'/*.jpg')
all_sizes = np.array([os.stat(t).st_size for t in all_tiles])

len(all_tiles)
```

```python
samples = [6, 18, 4, 31, 25, 61, 75, 80, 76, 137, 125, 159, 170, 186, 195, 203, 220, 224, 210]
plt.figure(figsize=(15,5))

n_cols = 7
for i, sample in enumerate(samples[:n_cols]):
    img_0_path = all_tiles[sample]
    img_0 = plt.imread(img_0_path)
    ax = plt.subplot(3,n_cols,i+1)
    ax.imshow(img_0)
        # 2. Hide all spines (the box)
    for side in ['top', 'right', 'bottom', 'left']:
        ax.spines[side].set_visible(False)
    # 3. Hide ticks and tick labels
    ax.tick_params(axis='both', which='both', length=0, labelbottom=False, labelleft=False)
    if i == 0:
        ax.set_ylabel('Original', fontsize=12)

    img_macenko_path = img_0_path.replace(original_path,macenko_path).replace('.jpg','_macenko.jpg')
    img_mac = plt.imread(img_macenko_path)
    ax = plt.subplot(3,n_cols,i+1+n_cols)
    ax.imshow(img_mac)
    # ax.axis('off')
        # 2. Hide all spines (the box)
    for side in ['top', 'right', 'bottom', 'left']:
        ax.spines[side].set_visible(False)
    # 3. Hide ticks and tick labels
    ax.tick_params(axis='both', which='both', length=0, labelbottom=False, labelleft=False)

    if i == 0:
        ax.set_ylabel('Macenko', fontsize=12)

    tile_id = img_0_path.rsplit('/',1)[-1]
    img_lut_path = LUT_path + slide + '/' + tile_id.replace('.jpg','.png')
    img_LUT = plt.imread(img_lut_path)
    ax = plt.subplot(3,n_cols,i+1+2*n_cols)
    ax.imshow(img_LUT)
        # 2. Hide all spines (the box)
    for side in ['top', 'right', 'bottom', 'left']:
        ax.spines[side].set_visible(False)
    # 3. Hide ticks and tick labels
    ax.tick_params(axis='both', which='both', length=0, labelbottom=False, labelleft=False)
    if i == 0:
        ax.set_ylabel('LUT', fontsize=12)

plt.tight_layout()
plt.show()
```
